# 프로젝트 3: Virtual Metrology 및 분류 기반 이상탐지 목표

PVD 데이터를 이용한 두께 회귀와 SECOM 데이터를 이용한 분류 기반 이상탐지를 각각 수행하고 결과를 비교합니다.

## 목표 1: 두께 Virtual Metrology

**두께 VM에는 PVD 데이터가 적합합니다.** 설비 센서값으로 평균 두께 또는 17개 지점의 두께를 예측하고, 다중회귀·트리 기반 모델 등을 비교할 수 있습니다. 평가에는 MAE, RMSE와 목표 규격 대비 오차를 사용합니다. 실제 규격과 두께 단위가 확인되지 않으면 분석용 가정과 데이터의 원래 스케일을 명시합니다.

## 목표 2: 분류 기반 이상탐지

SECOM의 공정 센서 특성과 합격·불합격 라벨을 이용해 분류 모델을 만들고 주요 변수를 선정합니다. 결측값과 불균형을 고려해 불량 재현율, 정밀도, F1 및 PR-AUC 등을 평가하고 하이퍼파라미터를 조정합니다. SECOM의 라벨은 두께 실측값이 아닙니다.

## 사용할 데이터와 주의점

- **PVD AlCu:** `../datasets/pvd_alcu/X_pvd_AlCu.csv`, `../datasets/pvd_alcu/Y_pvd_AlCu.csv` — [원본 데이터셋](https://zenodo.org/records/16881338). X·Y 행의 대응 기준과 측정값 단위 및 변환 방식을 확인해야 합니다.
- **SECOM:** `../datasets/secom/secom.data`, `../datasets/secom/secom_labels.data` — [UCI 원본 데이터셋](https://archive.ics.uci.edu/dataset/179/secom). 결측값과 적은 불합격 사례를 고려합니다.

두 데이터셋은 서로 다른 공정과 목표를 다루므로 회귀와 분류 점수를 직접적인 우열로 비교하지 않습니다. 데이터 분할과 전처리는 평가 데이터 누출을 피하도록 설계합니다.
