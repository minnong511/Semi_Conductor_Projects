# 프로젝트 2: Recipe 최적화 목표

프로젝트 1에서 사용한 PVD AlCu 데이터를 확장해 공정 조건 최적화를 탐색합니다.

## 목표

다중회귀분석과 최적화 알고리즘을 활용해 공정 변수와 두께·균일도의 관계를 분석합니다. 모델의 예측값을 이용해 **목표 두께 오차와 균일도를 함께 최소화하는 후보 조건**을 탐색할 수 있습니다. 공정능력과 예상 부작용도 함께 검토합니다.

## 데이터와 해석 범위

입력 데이터 `../datasets/pvd_alcu/X_pvd_AlCu.csv`와 17개 지점 두께 데이터 `../datasets/pvd_alcu/Y_pvd_AlCu.csv`를 사용합니다. [원본 데이터셋](https://zenodo.org/records/16881338)

센서 이름은 익명화되어 있으며, 센서 측정값이 실제로 조절 가능한 Recipe 설정값인지 확인되지 않았습니다. X와 Y의 행 대응 기준, 측정값 단위 및 변환 방식도 확인해야 합니다. 목표 두께와 규격을 가정한다면 명시하고, 공정능력 산정에 사용한 변동 추정과 균일도 정의를 함께 제시합니다. 탐색 결과는 추가 실험과 검증이 필요한 후보입니다.
